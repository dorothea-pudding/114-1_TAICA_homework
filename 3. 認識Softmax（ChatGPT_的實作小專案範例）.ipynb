```python
#匯入套件
import numpy as np
from ipywidgets import interact
import matplotlib.pyplot as plt
```

```python
##Softmax函式
def softmax(x):
    exp_x = np.exp(x - np.max(x))  # 減去最大值避免 overflow
    return exp_x / np.sum(exp_x)
```

```python
#骰子的加權機率
@interact(a=(-10, 10, 1), b=(-10, 10, 1), c=(-10, 10, 1),
          d=(-10, 10, 1), e=(-10, 10, 1), f=(-10, 10, 1))
def show_dice(a=1, b=1, c=1, d=1, e=1, f=1):
    scores = np.array([a, b, c, d, e, f])
    probs = softmax(scores)

    print("⚀⚁⚂⚃⚄⚅")
    print(f"輸入分數: {scores}")
    print(f"Softmax 機率: {probs.round(3)}")

    # 畫出長條圖
    plt.bar(["⚀","⚁","⚂","⚃","⚄","⚅"], probs)
    plt.title("加權骰子的機率分布")
    plt.ylim(0,1)
    plt.show()
```

```python
def roll_dice(scores, n=1000):
    probs = softmax(scores)
    outcomes = np.random.choice(["⚀","⚁","⚂","⚃","⚄","⚅"], size=n, p=probs)

    values, counts = np.unique(outcomes, return_counts=True)
    plt.bar(values, counts/n)
    plt.title(f"{n} 次擲骰子的實驗結果")
    plt.ylim(0,1)
    plt.show()

# 測試範例：讓 ⚅ 偏好分數很高
scores = np.array([2, 1, 7, 3, 1, 10])#骰子點數1~6的偏好分數
roll_dice(scores, n=5000)
```
