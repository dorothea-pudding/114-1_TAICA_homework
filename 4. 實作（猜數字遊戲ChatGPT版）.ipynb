給ChatGPT的指令：幫我用python寫一個猜數字遊戲，題目範圍1~100。

```python
import random

def guess_number():
    # 電腦隨機選一個 1 到 100 的整數
    answer = random.randint(1, 100)
    attempts = 0

    print("🎲 猜數字遊戲開始！範圍是 1 ~ 100")

    while True:
        try:
            guess = int(input("請輸入你的猜測："))
        except ValueError:
            print("⚠️ 請輸入有效的整數！")
            continue

        attempts += 1

        if guess < 1 or guess > 100:
            print("⚠️ 數字超出範圍，請重新輸入！")
        elif guess < answer:
            print("太小了！再大一點～")
        elif guess > answer:
            print("太大了！再小一點～")
        else:
            print(f"🎉 恭喜答對！答案就是 {answer}，總共猜了 {attempts} 次。")
            break

# 執行遊戲
guess_number()
```

給ChatGPT的指令：幫我改變程式，讓電腦隨機猜。我要它用random自己出題，然後它自己猜。

```python
import random

def computer_self_guess():
    # 電腦自己出題
    answer = random.randint(1, 100)
    attempts = 0
    guess = None

    print("🎲 電腦自己出題，自己亂猜 (1~100)！")

    while guess != answer:
        guess = random.randint(1, 100)  # 電腦亂猜
        attempts += 1
        print(f"第 {attempts} 次猜：{guess}")

    print(f"\n🎉 電腦終於猜對了！答案是 {answer}，總共猜了 {attempts} 次。")

# 執行
computer_self_guess()
```

給ChatGPT的指令：一樣的條件，幫我讓電腦在猜的時候使用二分法。

```python
import random

def computer_binary_guess():
    # 電腦自己出題
    answer = random.randint(1, 100)
    attempts = 0
    low, high = 1, 100
    guess = None

    print("🎲 電腦自己出題，自己用二分法來猜 (1~100)！")
    print(f"(秘密答案已隨機產生，電腦不知道答案是 {answer})\n")

    while guess != answer:
        # 二分法：取中間
        guess = (low + high) // 2
        attempts += 1
        print(f"第 {attempts} 次猜：{guess}")

        if guess < answer:
            print("➡️ 猜太小，縮小範圍")
            low = guess + 1
        elif guess > answer:
            print("⬅️ 猜太大，縮小範圍")
            high = guess - 1
        else:
            print(f"\n🎉 電腦猜對了！答案是 {answer}，總共猜了 {attempts} 次。")

# 執行
computer_binary_guess()
```
