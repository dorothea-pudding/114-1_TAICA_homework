給Gemini的指令：幫我用python寫一個猜數字遊戲，題目範圍1~100。

```python
import random

def guess_the_number():
    # 1. 生成一個1到100之間的隨機整數作為答案
    #    'random.randint(a, b)' 會包含 a 和 b 兩個端點。
    secret_number = random.randint(1, 100)

    # 初始化猜測次數和猜測值
    attempts = 0
    guess = None

    print("--- 猜數字遊戲開始！---")
    print("我心裡已經選好了一個 1 到 100 之間的數字。")
    print("你只有 7 次猜測機會，祝你好運！")
    print("-------------------------")

    # 設定最大猜測次數
    max_attempts = 7

    # 迴圈會持續進行，直到猜對為止，或者達到最大嘗試次數
    while guess != secret_number and attempts < max_attempts:
        try:
            # 2. 提示使用者輸入猜測的數字
            guess = int(input(f"第 {attempts + 1} 次猜測 (還剩 {max_attempts - attempts} 次機會): 請輸入你的數字: "))

            # 3. 檢查輸入的數字是否在範圍內
            if not (1 <= guess <= 100):
                print("請輸入一個 1 到 100 之間的數字！")
                continue # 跳過本次迴圈，重新開始

            attempts += 1

            # 4. 比較猜測值與答案
            if guess < secret_number:
                print("太低了！再往上猜一點。")
            elif guess > secret_number:
                print("太高了！再往下猜一點。")

        except ValueError:
            # 處理非數字輸入的錯誤
            print("無效的輸入。請輸入一個整數數字！")
            continue # 跳過本次迴圈，重新開始

    # 5. 遊戲結束後的結果判斷
    if guess == secret_number:
        print(f"\n恭喜你！你猜對了！答案就是 {secret_number}。")
        print(f"你總共猜了 {attempts} 次。")
    else:
        print(f"\n很可惜，你的 {max_attempts} 次機會用完了。")
        print(f"正確答案是 {secret_number}。下次再試試吧！")

# 執行遊戲
if __name__ == "__main__":
    guess_the_number()
```

給Gemini的指令：幫我改變程式，讓電腦隨機猜。我要它用random自己出題，然後它自己猜。

```python
import random

def computer_guesses_the_number():
    # 1. 電腦隨機設定一個 1 到 100 之間的秘密數字（答案）
    secret_number = random.randint(1, 100)

    # 初始化猜測範圍：從 1 到 100
    low = 1
    high = 100

    attempts = 0
    guess = None

    print("--- 電腦猜數字遊戲開始！---")
    print(f"我 (電腦) 心裡選了一個 1 到 100 之間的數字：{secret_number}")
    print("現在，我要用二分搜尋法來猜出這個數字！")
    print("---------------------------------")

    # 迴圈會持續進行，直到電腦猜對為止
    while guess != secret_number:
        # 確保猜測範圍有效
        if low > high:
            # 理論上二分搜尋不會發生，但作為錯誤檢查
            print("程式邏輯出錯了，範圍錯誤。")
            break

        # 2. 核心：使用二分搜尋法猜測中間值
        #    (low + high) // 2 取中間值，並向下取整
        guess = (low + high) // 2
        attempts += 1

        print(f"第 {attempts} 次猜測：電腦猜 {guess}")

        # 3. 判斷猜測結果並調整範圍
        if guess < secret_number:
            print(f"太低了！答案在 {guess + 1} 到 {high} 之間。")
            # 調整下限：答案一定在猜測值之上
            low = guess + 1

        elif guess > secret_number:
            print(f"太高了！答案在 {low} 到 {guess - 1} 之間。")
            # 調整上限：答案一定在猜測值之下
            high = guess - 1

    # 4. 遊戲結束
    print("---------------------------------")
    print(f"🎉 猜對了！電腦猜出答案是 {secret_number}。")
    print(f"電腦總共只花了 {attempts} 次就猜中了！")

# 執行電腦猜數字遊戲
if __name__ == "__main__":
    computer_guesses_the_number()
```
