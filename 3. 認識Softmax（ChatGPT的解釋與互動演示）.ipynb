# 🔢 Softmax 函式與「贏者通吃」效果
在這份 Notebook 中，我們將自己寫一個 **Softmax 函式**，並且用互動式介面觀察不同數字輸入後的結果。  
Softmax 的特性之一是：數字中最大的值會因為 **指數放大 (exponential scaling)**，而得到非常高的機率，其他值就會被壓得很小，呈現所謂的「贏者通吃」現象。

```python
# 📌 先載入需要的套件
import numpy as np
from ipywidgets import interact
#ipywidgets：為jupyter notebook 設計的互動式網頁元件庫
#interact：ipywidgets中用於自動產生調整參數的UI介面函數
import matplotlib.pyplot as plt  #視覺化數據函式庫
```

## 🧮 Softmax 函式
我們自己實作一個簡單的 Softmax 函式：  
公式為  

\[\text{softmax}(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}\]

其中：
- 分子：把每個輸入值取指數（會放大差異）
- 分母：做歸一化，讓所有結果加總為 1

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x))  # 減掉最大值，避免 overflow
    return exp_x / np.sum(exp_x)
```

## 🎛️ 互動觀察
我們用 `interact` 讓你可以自己輸入三個數字，觀察 Softmax 的輸出。  
請試著調整數字差距，看看結果是不是「最大值幾乎全拿」？

```python
#裝飾器語法：@寫在def上，代表將def放進裝飾器加工
@interact(a=(-10, 10, 1), b=(-10, 10, 1), c=(-10, 10, 1))
def show_softmax(a=0, b=0, c=0):
    x = np.array([a, b, c])
    result = softmax(x)
    print(f"輸入: {x}")
    print(f"Softmax 結果: {result}")
    print(f"總和: {np.sum(result)}")

    # 顯示圖形
    plt.bar(["a", "b", "c"], result)
    plt.title("Softmax 機率分布")
    plt.ylim(0, 1)
    plt.show()
```
