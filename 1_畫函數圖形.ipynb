```python
#引入套件
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
#在線上顯示圖形(預設另開視窗)
```

```python
a, b = 0, 0 #圓心座標
r = 3 #設定半徑
theta = np.linspace(0, 2*np.pi, 800) #產生角度(起始, 角度, 分割點數)
#算圓的描點座標
x = a + r * np.cos(theta)
y = b + r * np.sin(theta)
#x = np.linspace(-3*np.pi, 3*np.pi, 800) (函數起點, 函數終點, 描點個數) 這裡原本用來畫sin圖形，但是畫圓用不到
#y = np.log(x) #函數
```
